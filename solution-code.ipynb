```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
titanic = pd.read_csv('datasets/titanic_clean.csv')
```

```python
titanic.head()
```

### Student Preprocessing Exercise
Create my data matrix `X` and response vector `y` to meet the following specifications:
* Don't use `PassengerId` or `Name`
* One-hot encode and binarize where necessary

No need to use any fancy `sklearn` classes here, just plain `pandas` is fine.

```python
# Have at it!
X = pd.get_dummies(
    columns=['Pclass', 'Sex', 'Embarked'],
    data=titanic.drop(columns=['PassengerId', 'Name', 'Survived']),
    drop_first=True
)

y = titanic['Survived']
```

```python
X.head()
```

### Let's skip train-test splitting for now...
... why? Well, this is our last **white box model**! For the last time, we're not going to care about optimizing pure predictive ability. We're going to be able to peek into our white box and look at our model directly.

### Let's examine the `DecisionTreeClassifier` documentation:
Check it out [here](https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html)!

```python
DecisionTreeClassifier()
```

### Ok, cool, now let's fit one

```python
tree = DecisionTreeClassifier(max_depth=4, min_samples_split=5)
tree.fit(X, y)
```

```python
# We can see some feature importances...
tree.feature_importances_
```

```python
pd.DataFrame({
    'feature': X.columns,
    'importances': tree.feature_importances_
}).sort_values('importances', ascending=False)
```

### Visualizing a Decision Tree
(Sorry Windows users, this section might not be for you!)

```python
plt.figure(figsize=(16, 8))
plot_tree(
    tree,
    feature_names=X.columns,
    class_names=['Died', 'Survived'],
    filled=True
);
```

# Why use a decision tree?

### Pros:
* They're very white boxy
    * Easy to conceptualize
    * Easy to visualize
    * Easy to interpret (feature importances)
* They're nonparametric (no distributional assumptions)
* Scale of $x$s doesn't matter
    * Don't need to worry about skewness of feature or anything
    * Don't need to scale (more of a convenient fact than a pro)
* Speed
    * They're fast models to fit!
* Not a lot of feature engineering needed!
    * No need for interaction terms - the nature of the tree structure give us that interaction "and" logic automated for us!
    * Since the scale of our $x$s doesn't matter, we don't need to mess with polynomial terms.
* Handle categorical features and missing values tremendously
    * The yes/no logic of tree splits is perfectly designed for categorical features
    * Theoretically, if a row with a missing feature never makes it into a subset of the tree where that feature is used, we never needed to worry about that missing value!
        * Practically, `sklearn` still can't handle that situation though...

### Cons:
* Prone to overfitting without tuning
* Models tend to be overly simple when tuning
    * Yes you're reading that right... they're considered _both_ high bias _and_ high variance!
* Known to work poorly with imbalanced data
* CART algorithm is "greedy"
    * A **greedy** algorithm is one that makes the best decision without thinking one step in advance. They're often _fine_ algorithms but not _optimal_. It's totally possible the _best_ tree makes a foolish decision now only to have a better one later.

## Interview Question

<details><summary>If you're comparing decision trees and logistic regression, what are the pros and cons of each?</summary>
    
(Answers may vary; this is not an exhaustive list!)
- **Interpretability**: The coefficients in a logistic regression model are interpretable. (They represent the change in log-odds caused by the input variables.) However, this is complicated and not easy for non-technical audiences. Decision trees are interpretable; it is easy to explain to show a picture of a decision tree to a client or boss and get them to understand how predictions are made.
- **Performance**: Decision trees have a tendency to easily overfit, while logistic regression models usually do not overfit as easily.
- **Assumptions**: Decision trees have no assumptions about how data are distributed; logistic regression does make assumptions about how data are distributed.
- **Frequency**: Logistic regression is more commonly than decision trees.
- **Y variable**: Decision trees can handle regression and classification problems; logistic regression is only really used for classification problems.
